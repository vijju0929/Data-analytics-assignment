```python
# ============================================================
# SMART CROP RECOMMENDATION SYSTEM
# Data Analytics + Machine Learning Project
# Dataset: Kaggle Crop Recommendation Dataset
# Algorithm: Random Forest Classifier
# ============================================================

# -------------------------------
# 1. IMPORT LIBRARIES
# -------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import zipfile
import joblib
import json

from google.colab import files

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from sklearn.preprocessing import LabelEncoder

sns.set_style("whitegrid")

print("Libraries imported successfully!")


# ============================================================
# 2. UPLOAD KAGGLE DATASET
# ============================================================

print("Please upload your Kaggle Crop Recommendation CSV file.")

uploaded = files.upload()

# Automatically find uploaded CSV file
csv_file = list(uploaded.keys())[0]

df = pd.read_csv(csv_file)

print("\nDataset loaded successfully!")

# Display first five rows
display(df.head())


# ============================================================
# 3. BASIC DATA EXPLORATION
# ============================================================

print("\n========== DATASET INFORMATION ==========")

print("\nDataset Shape:")
print(df.shape)

print("\nNumber of Rows:")
print(df.shape[0])

print("\nNumber of Columns:")
print(df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# ============================================================
# 4. DATA CLEANING
# ============================================================

# Remove duplicate rows
df = df.drop_duplicates()

# Remove rows containing missing values
df = df.dropna()

print("\nDataset after cleaning:")
print(df.shape)

print("\nRemaining missing values:")
print(df.isnull().sum())

print("\nRemaining duplicate rows:")
print(df.duplicated().sum())


# ============================================================
# 5. IDENTIFY FEATURES AND TARGET
# ============================================================

features = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

target = "label"

X = df[features]
y = df[target]

print("\nInput Features:")
print(features)

print("\nTarget Variable:")
print(target)

print("\nNumber of Crop Categories:")
print(y.nunique())

print("\nCrop Categories:")
print(sorted(y.unique()))


# ============================================================
# 6. CREATE OUTPUT FOLDER
# ============================================================

output_folder = "Crop_Recommendation_Project"

os.makedirs(output_folder, exist_ok=True)

visualization_folder = os.path.join(
    output_folder,
    "visualizations"
)

os.makedirs(visualization_folder, exist_ok=True)


# ============================================================
# 7. VISUALIZATION 1 - CROP DISTRIBUTION
# ============================================================

plt.figure(figsize=(14, 7))

crop_counts = df["label"].value_counts()

sns.barplot(
    x=crop_counts.index,
    y=crop_counts.values
)

plt.title(
    "Distribution of Crop Categories",
    fontsize=16
)

plt.xlabel("Crop")
plt.ylabel("Number of Samples")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_folder,
        "01_crop_distribution.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 8. VISUALIZATION 2 - FEATURE DISTRIBUTIONS
# ============================================================

df[features].hist(
    figsize=(16, 12),
    bins=20,
    edgecolor="black"
)

plt.suptitle(
    "Distribution of Agricultural Features",
    fontsize=18
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_folder,
        "02_feature_distributions.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 9. VISUALIZATION 3 - CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(12, 8))

correlation_matrix = df[features].corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title(
    "Correlation Heatmap of Agricultural Features",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_folder,
        "03_correlation_heatmap.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 10. VISUALIZATION 4 - BOX PLOTS
# ============================================================

plt.figure(figsize=(16, 10))

for i, feature in enumerate(features, 1):

    plt.subplot(3, 3, i)

    sns.boxplot(
        x=df[feature]
    )

    plt.title(feature)

plt.suptitle(
    "Boxplots of Agricultural Features",
    fontsize=18
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_folder,
        "04_boxplots.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 11. VISUALIZATION 5 - FEATURE VS CROP
# ============================================================

plt.figure(figsize=(16, 8))

sns.boxplot(
    data=df,
    x="label",
    y="rainfall"
)

plt.title(
    "Rainfall Distribution Across Different Crops",
    fontsize=16
)

plt.xlabel("Crop")
plt.ylabel("Rainfall")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_folder,
        "05_rainfall_by_crop.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 12. VISUALIZATION 6 - TEMPERATURE VS HUMIDITY
# ============================================================

plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=df,
    x="temperature",
    y="humidity",
    hue="label",
    palette="tab20",
    alpha=0.7
)

plt.title(
    "Temperature vs Humidity for Different Crops",
    fontsize=16
)

plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")

plt.legend(
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_folder,
        "06_temperature_humidity_scatter.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 13. ENCODE TARGET LABELS
# ============================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("\nCrop Label Encoding:")
for crop, encoded_value in zip(
    label_encoder.classes_,
    range(len(label_encoder.classes_))
):
    print(crop, "=", encoded_value)


# ============================================================
# 14. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("\n========== TRAIN TEST SPLIT ==========")

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])


# ============================================================
# 15. TRAIN RANDOM FOREST MODEL
# ============================================================

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    max_depth=None,
    n_jobs=-1
)

print("\nTraining Random Forest Model...")

model.fit(
    X_train,
    y_train
)

print("Model training completed!")


# ============================================================
# 16. MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(X_test)

print("\nPredictions completed!")


# ============================================================
# 17. MODEL EVALUATION
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted"
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted"
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted"
)

print("\n========== MODEL PERFORMANCE ==========")

print("Accuracy:", round(accuracy * 100, 2), "%")

print("Precision:", round(precision * 100, 2), "%")

print("Recall:", round(recall * 100, 2), "%")

print("F1 Score:", round(f1 * 100, 2), "%")


# ============================================================
# 18. CLASSIFICATION REPORT
# ============================================================

print("\n========== CLASSIFICATION REPORT ==========")

report = classification_report(
    y_test,
    y_pred,
    target_names=label_encoder.classes_
)

print(report)


# ============================================================
# 19. VISUALIZATION 7 - CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(16, 14))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=label_encoder.classes_,
    yticklabels=label_encoder.classes_
)

plt.title(
    "Confusion Matrix - Random Forest Classifier",
    fontsize=18
)

plt.xlabel("Predicted Crop")
plt.ylabel("Actual Crop")

plt.xticks(rotation=45)
plt.yticks(rotation=0)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_folder,
        "07_confusion_matrix.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 20. VISUALIZATION 8 - MODEL PERFORMANCE METRICS
# ============================================================

metrics = {
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1
}

plt.figure(figsize=(10, 6))

metric_names = list(metrics.keys())
metric_values = list(metrics.values())

bars = plt.bar(
    metric_names,
    metric_values
)

plt.title(
    "Random Forest Model Performance",
    fontsize=16
)

plt.ylabel("Score")
plt.ylim(0, 1.1)

for bar, value in zip(bars, metric_values):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.02,
        f"{value:.3f}",
        ha="center",
        fontsize=12
    )

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_folder,
        "08_model_performance_metrics.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 21. VISUALIZATION 9 - FEATURE IMPORTANCE
# ============================================================

importance = model.feature_importances_

feature_importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": importance
})

feature_importance_df = feature_importance_df.sort_values(
    by="Importance",
    ascending=False
)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=feature_importance_df,
    x="Importance",
    y="Feature"
)

plt.title(
    "Feature Importance in Crop Recommendation",
    fontsize=16
)

plt.xlabel("Importance Score")
plt.ylabel("Agricultural Feature")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_folder,
        "09_feature_importance.png"
    ),
    dpi=300
)

plt.show()

print("\nFeature Importance:")
display(feature_importance_df)


# ============================================================
# 22. CROSS VALIDATION
# ============================================================

print("\n========== CROSS VALIDATION ==========")

cv_scores = cross_val_score(
    model,
    X,
    y_encoded,
    cv=5,
    scoring="accuracy"
)

print("5-Fold Cross Validation Scores:")
print(cv_scores)

print(
    "Average Cross Validation Accuracy:",
    round(cv_scores.mean() * 100, 2),
    "%"
)

print(
    "Standard Deviation:",
    round(cv_scores.std() * 100, 2),
    "%"
)


# ============================================================
# 23. SAMPLE CROP PREDICTION
# ============================================================

print("\n========== SAMPLE CROP PREDICTION ==========")

# Sample agricultural input values
sample_data = pd.DataFrame(
    [[
        90,       # Nitrogen
        42,       # Phosphorus
        43,       # Potassium
        20.8,     # Temperature
        82.0,     # Humidity
        6.5,      # pH
        202.9     # Rainfall
    ]],
    columns=features
)

predicted_crop_encoded = model.predict(
    sample_data
)

predicted_crop = label_encoder.inverse_transform(
    predicted_crop_encoded
)

print("\nSample Input:")
display(sample_data)

print("\nRecommended Crop:")
print(predicted_crop[0].upper())


# ============================================================
# 24. CREATE A SIMPLE PREDICTION FUNCTION
# ============================================================

def recommend_crop(
    nitrogen,
    phosphorus,
    potassium,
    temperature,
    humidity,
    ph,
    rainfall
):

    input_data = pd.DataFrame(
        [[
            nitrogen,
            phosphorus,
            potassium,
            temperature,
            humidity,
            ph,
            rainfall
        ]],
        columns=features
    )

    prediction = model.predict(input_data)

    crop = label_encoder.inverse_transform(
        prediction
    )[0]

    probabilities = model.predict_proba(
        input_data
    )[0]

    confidence = max(probabilities) * 100

    return crop, confidence


# Example
crop_result, confidence_result = recommend_crop(
    nitrogen=90,
    phosphorus=42,
    potassium=43,
    temperature=20.8,
    humidity=82,
    ph=6.5,
    rainfall=202.9
)

print("\nCrop Recommendation:", crop_result.upper())

print(
    "Prediction Confidence:",
    round(confidence_result, 2),
    "%"
)


# ============================================================
# 25. SAVE MODEL
# ============================================================

joblib.dump(
    model,
    os.path.join(
        output_folder,
        "random_forest_crop_model.pkl"
    )
)

joblib.dump(
    label_encoder,
    os.path.join(
        output_folder,
        "crop_label_encoder.pkl"
    )
)

print("\nModel saved successfully!")


# ============================================================
# 26. SAVE EVALUATION RESULTS
# ============================================================

evaluation_results = {
    "Algorithm": "Random Forest Classifier",
    "Accuracy": float(accuracy),
    "Precision": float(precision),
    "Recall": float(recall),
    "F1 Score": float(f1),
    "Cross Validation Mean Accuracy": float(cv_scores.mean()),
    "Cross Validation Standard Deviation": float(cv_scores.std()),
    "Training Samples": int(X_train.shape[0]),
    "Testing Samples": int(X_test.shape[0]),
    "Number of Crop Categories": int(y.nunique())
}

with open(
    os.path.join(
        output_folder,
        "evaluation_results.json"
    ),
    "w"
) as file:

    json.dump(
        evaluation_results,
        file,
        indent=4
    )


# Save metrics as CSV
metrics_df = pd.DataFrame(
    [evaluation_results]
)

metrics_df.to_csv(
    os.path.join(
        output_folder,
        "model_evaluation_results.csv"
    ),
    index=False
)


# Save classification report
with open(
    os.path.join(
        output_folder,
        "classification_report.txt"
    ),
    "w"
) as file:

    file.write(
        classification_report(
            y_test,
            y_pred,
            target_names=label_encoder.classes_
        )
    )


# Save feature importance
feature_importance_df.to_csv(
    os.path.join(
        output_folder,
        "feature_importance.csv"
    ),
    index=False
)


# Save cleaned dataset
df.to_csv(
    os.path.join(
        output_folder,
        "cleaned_crop_dataset.csv"
    ),
    index=False
)


# ============================================================
# 27. CREATE PROJECT README
# ============================================================

readme_content = f"""
# Smart Crop Recommendation System

## Project Overview

This project uses Machine Learning to recommend a suitable crop
based on soil nutrient levels and climatic conditions.

## Objective

To predict the most suitable crop using agricultural parameters
such as nitrogen, phosphorus, potassium, temperature, humidity,
soil pH, and rainfall.

## Dataset

Kaggle Crop Recommendation Dataset.

## Input Features

- Nitrogen (N)
- Phosphorus (P)
- Potassium (K)
- Temperature
- Humidity
- Soil pH
- Rainfall

## Target Variable

Crop Label

## Machine Learning Algorithm

Random Forest Classifier

## Tools and Technologies

- Python
- Google Colab
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Scikit-learn
- Joblib

## Model Evaluation

Accuracy: {accuracy*100:.2f}%

Precision: {precision*100:.2f}%

Recall: {recall*100:.2f}%

F1 Score: {f1*100:.2f}%

Cross Validation Accuracy: {cv_scores.mean()*100:.2f}%

## Visualizations

The project includes:

1. Crop Distribution
2. Feature Distributions
3. Correlation Heatmap
4. Boxplots
5. Rainfall by Crop
6. Temperature vs Humidity
7. Confusion Matrix
8. Model Performance Metrics
9. Feature Importance

## Conclusion

The Random Forest model was trained to predict suitable crops
based on soil and environmental parameters. The model performance
was evaluated using accuracy, precision, recall, F1-score,
confusion matrix and cross-validation.

## Disclaimer

This project is an educational machine learning demonstration.
Crop recommendations should be validated with local agricultural
conditions, soil testing and expert agricultural advice.
"""

with open(
    os.path.join(
        output_folder,
        "README.md"
    ),
    "w"
) as file:

    file.write(readme_content)


# ============================================================
# 28. SAVE DATASET COPY
# ============================================================

df.to_csv(
    os.path.join(
        output_folder,
        "Crop_Recommendation_Dataset.csv"
    ),
    index=False
)


# ============================================================
# 29. CREATE ZIP FILE FOR GITHUB PROJECT
# ============================================================

zip_filename = "Smart_Crop_Recommendation_Project.zip"

with zipfile.ZipFile(
    zip_filename,
    "w",
    zipfile.ZIP_DEFLATED
) as zipf:

    for root, dirs, files_list in os.walk(output_folder):

        for file_name in files_list:

            file_path = os.path.join(
                root,
                file_name
            )

            arcname = os.path.relpath(
                file_path,
                output_folder
            )

            zipf.write(
                file_path,
                arcname
            )


# ============================================================
# 30. FINAL SUMMARY
# ============================================================

print("\n" + "="*60)

print("PROJECT COMPLETED SUCCESSFULLY!")

print("="*60)

print("\nPROJECT: Smart Crop Recommendation System")

print("\nALGORITHM: Random Forest Classifier")

print("\nMODEL PERFORMANCE:")

print("Accuracy:", round(accuracy*100, 2), "%")

print("Precision:", round(precision*100, 2), "%")

print("Recall:", round(recall*100, 2), "%")

print("F1 Score:", round(f1*100, 2), "%")

print(
    "Cross Validation Accuracy:",
    round(cv_scores.mean()*100, 2),
    "%"
)

print("\nFiles created:")

print("- README.md")
print("- Cleaned dataset")
print("- Trained ML model")
print("- Label encoder")
print("- Evaluation metrics")
print("- Classification report")
print("- Feature importance")
print("- 9 visualizations")

print("\nDownloading project ZIP file...")

files.download(zip_filename)

print("\nDownload complete!")
print("You can now upload the project ZIP contents to GitHub.")
```
